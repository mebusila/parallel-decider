# V0 — Single Boolean Decision

This notebook verifies the first end-to-end decision flow:

1. Load a local NLI model.
2. Define a natural-language state.
3. Define one Boolean hypothesis.
4. Return the entailment probability as a `BooleanDecision`.

This is intentionally sequential and will serve as the baseline before batched inference.

```python
from parallel_decider import BooleanQuestion, NLIBackend

backend = NLIBackend()
```

```python
state = "Find which commit introduced the authentication bug."

questions = [
    BooleanQuestion(
        name="needs_git",
        hypothesis="This task requires access to Git.",
    )
]
```

```python
result = backend.decide(
    state=state,
    questions=questions,
)

result
```

```python
for decision in result:
    print(f"{decision.name}: {decision.probability:.3f}")
```
